## What I Learned From the Exercises
### Exercises (thin ellipse, circle, rotated, shifted)
- For the **thin ellipse**, I observed stronger anisotropy, and that showed up as richer behavior in higher-order coefficients.
- For the **circle**, I saw that symmetry simplifies the representation, and higher-order terms decayed quickly.
- For the **rotated ellipse**, I learned that changing orientation mostly changes coefficient phase (real/imaginary structure), while the area-related term stays relatively stable.
- For the **shifted ellipse**, I saw that translation strongly affects lower-order moments, especially \(A_1\).
- This gave me practical intuition for interpreting model errors physically (size vs center vs orientation vs shape complexity).
### Connection to Minimal-Data Inverse Gravimetry Theory
- From the first meeting material, I understood that only the first few terms are stably recoverable in practice because of background uncertainty and noise.
- I also understood why the theory focuses on recovering ellipse parameters from a small number of measurements by solving for \(A_0, A_1, A_2\).
- In Task 1, I implemented the computational side of those coefficients first, before adding noisy measurement modeling.
- This helped me see why the project timeline is structured as: foundations first, then data generation/noise, then ML.
### Why My Theta/Area Bug Was Important
- I found that using an incorrect rotation formula made the apparent area change with \(\theta\), which should not happen physically.
- Fixing the rotation matrix was important because a correct rotation should preserve area (up to small grid discretization effects).
- Catching this early was a key validation step and prevents corrupted labels/coefficient targets later in the pipeline.
- Rotating an object can change its area (negligible) based on how many pixels we give it
