## Gaussian Noise Basics

Gaussian noise is random noise with a normal distribution, written as `N(0, sigma^2)`.

Key ideas:
- It is additive: `y = x + n`.
- It is independent from sample to sample.
- Positive and negative errors are equally likely.
- Most values stay near 0, but large values can still happen.

For complex data, the real and imaginary parts are noised separately.

## Demo 1 Notes

The first demo adds Gaussian noise to complex gradient data.
The main point is that the same clean signal can look very different as `sigma` increases.

The helper functions do three things:
- `addGaussianNoise`: adds noise to the real and imaginary parts
- `addGaussianNoiseDetailed`: also returns the noise components
- `estimateNoiseLevel`: gives a rough noise estimate from the noisy signal

Insight: once `sigma` is large enough, the noisy gradient is no longer a small perturbation.
It can change the reconstruction path in a meaningful way.

## Demo 2 Notes

This demo shows the effect of noise visually in two ways:
- on the complex plane
- as an error magnitude plot

What to look for as noise grows:
- points move farther away from the clean signal
- the scatter becomes less structured
- the absolute error values get larger

Practical takeaway: visual checks help confirm whether the noise level is mild, moderate, or severe before reconstruction.

## Robustness Experiment Notes

The experiment varies both the number of measurement points `N` and the noise level `sigma`.

Expected trend:
- larger `sigma` usually lowers IoU and raises MSE
- larger `N` usually helps because the reconstruction gets more information
- small `N` is more sensitive to noise because each point matters more

Insight: the best results usually come from a balance of enough samples and low enough noise.
If noise is too strong, adding more points helps but cannot fully recover lost signal.

## Metrics Notes

The PDF uses three important metrics:

- IoU: measures overlap between predicted and true masks. Higher is better.
- MSE: measures average squared error. Lower is better.
- SNR: compares useful signal power to noise power in dB. Higher is better.

Quick interpretation:
- high IoU means the predicted shape matches the true shape well
- low MSE means the mask values are close to the ground truth
- high SNR means the measurements are cleaner

Note: IoU is best for shape overlap, while MSE is better for pixel-wise numeric error.

## Main Takeaway

Task 5 shows that Gaussian noise affects both the measurements and the final reconstruction quality.
The reconstruction is more reliable when the gradient data is cleaner, the sampling is sufficient, and the evaluation uses the right metric for the job.

Short answer for the task question: training or reconstruction usually takes longer and works less well as noise increases, because the model has to recover the signal from more uncertainty.
