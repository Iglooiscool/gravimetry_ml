# Task 4 Notes

These are short answers and observations for the student exercises in the Task 4 PDF.
I wrote them in simple language so they can be used as study notes.

## Exercise 1

Train the overfitting network from Demo 1 and observe the validation loss curve.
At what epoch does overfitting begin?

In general, overfitting starts when the training loss keeps going down but the validation
loss stops improving and begins to rise. In this kind of demo, that often happens after
the first several dozen epochs, but the exact epoch can change a little from run to run.

## Exercise 2

Compare dropout rates of 0.2, 0.5, and 0.8. What happens with very high dropout?

A small dropout rate like 0.2 usually helps without hurting learning too much.
A medium rate like 0.5 is stronger and can still work well if the network is large enough.
A very high rate like 0.8 usually removes too much information, so the model starts
underfitting and both training and validation performance get worse.

## Exercise 3

Implement L2 regularization with different lambda values (0, 0.0001, 0.001, 0.01).

- `0` means no penalty, so the network can overfit more easily.
- `0.0001` is a light penalty and often gives a small improvement.
- `0.001` is a moderate penalty and is usually a good starting point.
- `0.01` is stronger and may help if the model is very noisy, but it can also start to underfit if it is too large.

## Main Takeaway

The main lesson from Task 4 is that overfitting is not fixed by one trick only.
The best result usually comes from combining the right model size with early stopping,
dropout, and L2 regularization.
