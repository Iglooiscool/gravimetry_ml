## Why Loss Functions Matter

A neural network does not improve unless it has a number to minimize.
The loss function gives that number.

If the predictions are good, the loss should be small.
If the predictions are bad, the loss should be large.

## DNN vs MLP

A DNN is a general deep neural network with multiple hidden layers.
An MLP is a feedforward type of DNN.

Main parts:
- input layer
- one or more hidden layers
- output layer

## Binary Classification Setup

Binary classification means there are only two classes, usually written as `0` and `1`.

Typical output choice:
- one output neuron
- sigmoid activation or an equivalent logits-based loss

## Sigmoid Intuition

The sigmoid function maps any real number into the range `0` to `1`.
That makes it useful for probabilities.

If the output is near `1`, the model is confident in class `1`.
If the output is near `0`, the model is confident in class `0`.

## Why BCE Fits Classification

Binary cross-entropy is designed for probability outputs in binary classification.
It punishes confident wrong answers strongly.

That is why BCE is usually better than MSE for this task.
MSE can work, but it often learns more slowly for wrong predictions.

## PyTorch Notes

In PyTorch, the most common choice is `nn.BCEWithLogitsLoss()`.
It combines the sigmoid step and the BCE calculation into one numerically stable loss.

If you already have probabilities after a sigmoid, you can use `nn.BCELoss()`, but that is usually not the preferred training path.

## Main Takeaway

The big lesson is simple: use a loss function that matches the problem.
For binary classification, BCE is usually the right default, and in PyTorch the safest implementation is `BCEWithLogitsLoss`.
